- time needs to be in minutes or hours

```python
import re
def to_minutes(x):
    x = str(x).lower()
    if "less than 1 min" in x:
        return 0
    
    # Extract hours and minutes using regex
    h_match = re.search(r'(\d+)\s*hr', x)
    m_match = re.search(r'(\d+)\s*min', x)
    
    hours = int(h_match.group(1)) if h_match else 0
    minutes = int(m_match.group(1)) if m_match else 0
    
    return hours*60 + minutes
audible_df["time_mins"] = audible_df["time"].apply(to_minutes)
```

```python
audible_df.drop(columns='time',inplace=True)
audible_df.rename(columns={'time_mins':'time(mins)'},inplace=True)
```

- name in author and narrator there is no space between surname and first name

```python
def split_names(val):
    # Handle NaN or blank values
    if pd.isna(val) or str(val).strip() == "":
        return val  
    
    val = str(val)  # make sure it's a string
    parts = val.split(",")
    cleaned_parts = []
    for p in parts:
        # insert space before capital letters (CamelCase → Camel Case)
        name = re.sub(r'(?<!^)(?=[A-Z])', ' ', p).strip()
        cleaned_parts.append(name)
    return ", ".join(cleaned_parts)

audible_df["narrator"] = audible_df["narrator"].apply(split_names)
audible_df["author"] = audible_df["author"].apply(split_names)
```

```python
# replacing blank spaces in author and narrator column with no data avaliable
audible_df["author"] = audible_df["author"].replace(r'^\s*$', "No data available", regex=True)
audible_df["narrator"] = audible_df["narrator"].replace(r'^\s*$', "No data available", regex=True)
```
